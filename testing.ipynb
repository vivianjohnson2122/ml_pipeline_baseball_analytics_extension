```python
from pathlib import Path

mlruns = Path("./mlruns/0")
for run_dir in mlruns.iterdir():
    if not run_dir.is_dir() or run_dir.name.startswith("."):
        continue
    print(f"\n--- {run_dir.name[:12]} ---")
    
    # tags
    tags_dir = run_dir / "tags"
    if tags_dir.exists():
        for f in tags_dir.iterdir():
            print(f"  TAG {f.name}: {f.read_text().strip()}")
    
    # metrics
    metrics_dir = run_dir / "metrics"
    if metrics_dir.exists():
        for f in metrics_dir.iterdir():
            print(f"  METRIC {f.name}: {f.read_text().strip()}")
    
    # artifacts
    artifacts_dir = run_dir / "artifacts"
    if artifacts_dir.exists():
        for f in artifacts_dir.rglob("*"):
            print(f"  ARTIFACT {f.relative_to(run_dir)}")
    else:
        print("  NO ARTIFACTS DIR")
```

```python
from pathlib import Path

models_dir = Path("./mlruns/0/models")
for model_dir in sorted(models_dir.iterdir()):
    if not model_dir.is_dir():
        continue
    print(f"\n--- {model_dir.name} ---")
    
    tags_dir = model_dir / "tags"
    if tags_dir.exists():
        for f in tags_dir.iterdir():
            print(f"  TAG {f.name}: {f.read_text().strip()}")
    
    artifacts_dir = model_dir / "artifacts"
    if artifacts_dir.exists():
        for f in artifacts_dir.rglob("*"):
            print(f"  ARTIFACT {f.relative_to(model_dir)}")
    else:
        print("  NO ARTIFACTS")
```

```python
from pathlib import Path

models_dir = Path("./mlruns/0/models")
for model_dir in sorted(models_dir.iterdir()):
    if not model_dir.is_dir():
        continue
    artifacts = model_dir / "artifacts"
    files = [f.name for f in artifacts.iterdir()] if artifacts.exists() else ["NO ARTIFACTS"]
    print(f"{model_dir.name}: {files}")
```

```python
from pathlib import Path

exp_path = Path("./mlruns/0")
for run_dir in sorted(exp_path.iterdir()):
    if not run_dir.is_dir() or not all(c in "0123456789abcdef" for c in run_dir.name):
        continue
    outputs = run_dir / "outputs"
    if outputs.exists():
        links = [e.name for e in outputs.iterdir()]
        print(f"{run_dir.name[:12]}: outputs/ -> {links}")
    else:
        print(f"{run_dir.name[:12]}: NO outputs dir")
```
